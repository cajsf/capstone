## 다운로드 + 압축해제 + Drive 복사 함수

```
DRIVE_BASE = '/content/drive/MyDrive/capstone/DMD/distraction'
os.makedirs(DRIVE_BASE, exist_ok=True)

def process_one_file(url, idx, total):
    """한 파일을 다운로드 → 압축해제 → face+json만 Drive 복사 → 로컬 삭제"""
    filename = url.split('/')[-1].split('?')[0]
    print(f'\n{"="*60}')
    print(f'[{idx}/{total}] {filename}')
    print(f'{"="*60}')
    
    # 1. 다운로드 (로컬)
    tar_path = os.path.join(TEMP_DIR, filename)
    if os.path.exists(tar_path):
        os.remove(tar_path)
    print(f'  1) 다운로드 중...')
    start = time.time()
    !wget -q --show-progress -O "{tar_path}" "{url}"
    
    if not os.path.exists(tar_path) or os.path.getsize(tar_path) < 1000:
        print(f'  ❌ 다운로드 실패! URL 만료 가능성')
        if os.path.exists(tar_path):
            os.remove(tar_path)
        return False
    
    size_mb = os.path.getsize(tar_path) / (1024*1024)
    elapsed = time.time() - start
    print(f'  → 완료: {size_mb:.0f} MB ({elapsed:.0f}초)')
    
    # 2. 압축 해제 (로컬)
    extract_dir = os.path.join(TEMP_DIR, 'extracted')
    if os.path.exists(extract_dir):
        shutil.rmtree(extract_dir)
    os.makedirs(extract_dir)
    
    print(f'  2) 압축 해제 중...')
    start = time.time()
    try:
        with tarfile.open(tar_path, 'r:gz') as tar:
            tar.extractall(path=extract_dir)
    except Exception as e:
        print(f'  ❌ 압축 해제 실패: {e}')
        # tar.gz가 아닐 수도 있으니 그냥 tar로도 시도
        try:
            with tarfile.open(tar_path, 'r:') as tar:
                tar.extractall(path=extract_dir)
        except:
            print(f'  ❌ tar 해제도 실패. 건너뜀.')
            os.remove(tar_path)
            shutil.rmtree(extract_dir, ignore_errors=True)
            return False
    elapsed = time.time() - start
    print(f'  → 완료 ({elapsed:.0f}초)')
    
    # 3. tar.gz 삭제 (로컬 공간 확보)
    os.remove(tar_path)
    print(f'  3) tar.gz 삭제 완료')
    
    # 4. face 카메라 파일 + JSON만 Drive로 복사
    copied = 0
    skipped = 0
    print(f'  4) Drive로 복사 중...')
    start = time.time()
    for root, dirs, files in os.walk(extract_dir):
        for f in files:
            # face 카메라 파일 + annotation JSON만 복사
            if 'face' in f.lower() or f.endswith('.json'):
                src = os.path.join(root, f)
                rel_path = os.path.relpath(src, extract_dir)
                dst = os.path.join(DRIVE_BASE, rel_path)
                
                # 이미 존재하면 건너뜀
                if os.path.exists(dst) and os.path.getsize(dst) > 0:
                    skipped += 1
                    continue
                
                os.makedirs(os.path.dirname(dst), exist_ok=True)
                shutil.copy2(src, dst)
                copied += 1
            else:
                skipped += 1
    elapsed = time.time() - start
    print(f'  → 복사: {copied}개, 건너뜀: {skipped}개 ({elapsed:.0f}초)')
    
    # 5. 로컬 정리
    shutil.rmtree(extract_dir)
    print(f'  5) 로컬 정리 완료 ✅')
    return True

print(f'Drive 저장 경로: {DRIVE_BASE}')
print(f'로컬 임시 경로: {TEMP_DIR}')
```

## 실행

```
# === 다운로드 실행 ===
success = 0
fail = 0
total = len(URLS)

print(f'총 {total}개 파일 처리 시작!')
print(f'예상 소요 시간: 파일당 5~15분\n')

for i, url in enumerate(URLS, 1):
    ok = process_one_file(url, i, total)
    if ok:
        success += 1
    else:
        fail += 1

print(f'\n{"="*60}')
print(f'  완료! 성공: {success}/{total}, 실패: {fail}/{total}')
print(f'{"="*60}')
```

## 결과 확인

```
# === Drive에 저장된 파일 확인 ===
total_files = 0
total_size = 0
for root, dirs, files in os.walk(DRIVE_BASE):
    for f in files:
        fpath = os.path.join(root, f)
        total_files += 1
        total_size += os.path.getsize(fpath)

print(f'Drive 저장 경로: {DRIVE_BASE}')
print(f'총 파일 수: {total_files}')
print(f'총 크기: {total_size/(1024**3):.1f} GB')

# 폴더 구조 미리보기
print(f'\n=== 폴더 구조 (3단계까지) ===')
for root, dirs, files in os.walk(DRIVE_BASE):
    level = root.replace(DRIVE_BASE, '').count(os.sep)
    if level < 3:
        indent = '  ' * level
        print(f'{indent}{os.path.basename(root)}/')
        if level == 2:
            subindent = '  ' * (level + 1)
            face_files = [f for f in files if 'face' in f.lower()]
            json_files = [f for f in files if f.endswith('.json')]
            print(f'{subindent}face 비디오: {len(face_files)}개, annotation: {len(json_files)}개')
```
